```python
## Automate extract HTML Table for Jan 2025 - requires Selenium
## 1. Install latest Google Chrome browser
## 2. Download chromedriver from https://googlechromelabs.github.io/chrome-for-testing/
##    i. Scroll down to Stable section
##    ii. Locate chromedriver for win64
##    iii. Copy its http link, then Paste & Go in a new tab or new window - download the ZIP file
##    iv. Copy over the chromedriver.exe from the ZIP file into the folder containing this notebook
## 3. Manually inspect element to get values of the HTML attributes
##    i. tab "30 days"
##    ii. button "More days"
##    iii. aria-hidden="false"
##    iv. table class="bvYrdf"
## 4. DON'T FORGET to driver.quit(), or else the Google Chrome will kept open!
## Note: This tutorials is made possible via YouTube video - https://www.youtube.com/watch?v=NB8OceGZGjA
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.common.by import By

service = Service(executable_path="chromedriver.exe")
driver = webdriver.Chrome(service=service)
driver.get(url="https://www.google.com/search?q=prayer+times+in+kuala+lumpur")

button30Days = driver\
    .find_element(By.CLASS_NAME, "aYI8We")\
    .find_element(By.PARTIAL_LINK_TEXT, "30 days")
button30Days.click()

buttonMoreDays = driver\
    .find_element(By.TAG_NAME, "g-inline-expansion-bar")
buttonMoreDays.click()

htmlTable = driver\
    .find_element(By.CLASS_NAME, "Kp6KVb")\
    .find_element(By.XPATH, "//g-expandable-content[@aria-hidden='false']")\
    .find_element(By.TAG_NAME, "table")
htmlTableString = htmlTable.get_attribute("outerHTML")

driver.quit()
```

```python
import pandas as pd
from io import StringIO

df = pd.read_html(StringIO(str(htmlTableString)))[0]
df.iloc[0,0] = "2025"

df
```
